4. Create a random numpy array of size 6x3 and transpose it

```python
arr = np.random.rand(6,3)
transposed = arr.T
print("Original:")
print(arr)
print("Transposed:")
print(transposed)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product


```python
n = 2
p = 3
m = 4

A = np.random.rand(n, p)
B = np.random.rand(p, m)

print("Array A:")
print(A)

print("\nArray B:")
print(B)


result = np.dot(A, B)

print("\nDot product:")
print(result)
```
